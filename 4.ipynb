```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn as slice
```

```python
df.dropna(inplace=True)
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
loyal_cust=df[df['Customer Type']=='Loyal Customer']
disloyal_cust=df[df['Customer Type']=='disloyal Customer']
```

```python
loyal_customer=loyal_cust.sample(21000)
```

```python
loyal_cust.value_counts()
```

```python
loyal_customer.shape
```

```python
disloyal_cust.value_counts()
```

```python
balance_df=pd.concat([loyal_customer,disloyal_cust],axis=0)
balance_df.head()
```

```python
cat_col=balance_df.select_dtypes(include='O')
cat_col.head()
```

```python
num_col=balance_df.select_dtypes(exclude='O')
num_col.head()
```

```python
from sklearn.preprocessing import LabelEncoder,OneHotEncoder
```

```python
encoder= OneHotEncoder(drop='if_binary')
```

```python
encoder=OneHotEncoder()
```

```python
print(OneHotEncoder.__doc__)
```

```python
onehot_data=encoder.fit_transform(cat_col).toarray()
```

```python
onehot_data
```

```python
onehot_data.shape
```

```python
column_names=list(encoder.get_feature_names_out())
column_names
```

```python
onehot=pd.DataFrame(onehot_data,columns=column_names)
onehot
```

```python
final_df=pd.concat([onehot,num_col],axis=1)
final_df
```

```python
final_df.to_csv('cust_airline_cleaned.csv')
```

```python
x=final_df.drop(['Customer Type_disloyal Customer'],axis=1)
y=final_df['Customer Type_disloyal Customer']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.2)
```

```python
x_train
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression
lr.fit(x_train,y_train)
```

```python
lr.score(x_test,y_test)
```
